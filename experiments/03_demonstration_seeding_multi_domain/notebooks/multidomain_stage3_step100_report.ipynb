# Multi-domain Stage 3 — Step 100 completed-run report

Read-only: this notebook does not load a model, change a checkpoint, or train.


```python
from pathlib import Path
import json
from google.colab import drive

drive.mount('/content/drive', force_remount=False)
REPORT = Path('/content/drive/MyDrive/AISI/checkpoints/multidomain-stage3-v1/step100_fan_fix_report.json')
CHECKPOINT = Path('/content/drive/MyDrive/AISI/checkpoints/multidomain-stage3-v1/trainer-output/checkpoint-100')
if not REPORT.is_file():
    raise RuntimeError(f'Missing report: {REPORT}')
report = json.loads(REPORT.read_text())
if not report.get('step100_complete'):
    raise RuntimeError('The report exists but is not marked step100_complete.')
required = ('trainer_state.json', 'optimizer.pt', 'scheduler.pt')
checkpoint_ok = CHECKPOINT.is_dir() and all((CHECKPOINT / name).is_file() and (CHECKPOINT / name).stat().st_size > 0 for name in required)
adapter_ok = CHECKPOINT.is_dir() and any(p.name.startswith('adapter_model') and p.stat().st_size > 0 for p in CHECKPOINT.iterdir())
print('COMPLETED RUN VERIFIED:', {'step100_complete': True, 'checkpoint': str(CHECKPOINT), 'checkpoint_valid': checkpoint_ok and adapter_ok})
```

```python
print('===== STEP 100 PER-DOMAIN AND POOLED METRICS =====')
print(json.dumps(report.get('metrics', {}), indent=2, sort_keys=True))

print('===== FAN WORDING-FIX RESPONSE =====')
print(json.dumps(report.get('fan_response_trajectory', {}), indent=2, sort_keys=True))

print('===== VALVE/LAMP NON-REGRESSION =====')
print(json.dumps(report.get('valve_lamp_nonregression', {}), indent=2, sort_keys=True))

print('===== POOLED 95% ACCEPTANCE GATE =====')
print(json.dumps(report.get('pooled_acceptance_gate', {}), indent=2, sort_keys=True))
```

```python
metrics = report['metrics']
summary = {
    'checkpoint': report.get('checkpoint'),
    'stop_reason': report.get('stop_reason'),
    'fan_tracking': metrics['per_domain']['fan']['transition_tracking_rate'],
    'valve_tracking': metrics['per_domain']['valve']['transition_tracking_rate'],
    'lamp_tracking': metrics['per_domain']['lamp']['transition_tracking_rate'],
    'pooled_accuracy': metrics['pooled']['physical_final_answer_accuracy'],
    'full_95_percent_gate_passed': report['pooled_acceptance_gate']['full_pooled_gate_passed'],
    'stage4_authorized': report.get('stage4_authorized', False),
}
print('===== TL;DR =====')
print(json.dumps(summary, indent=2, sort_keys=True))
print('READ-ONLY INSPECTION COMPLETE. No training was run.')
```
